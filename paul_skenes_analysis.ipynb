```python
import sqlite3

# SQLiteデータベースを作成
conn = sqlite3.connect("skenes.db")
```

```python
import pandas as pd

# CSVファイルを読み込む
reference_2024 = pd.read_csv("2024_paul skines_data_reference.csv")
reference_2025 = pd.read_csv("2025_paul skines_data_reference.csv")

savant_2024 = pd.read_csv("savant_2024_paul skenes.csv")
savant_2025 = pd.read_csv("savant_2025_paul skenes.csv")

savant_pitch_type_2024 = pd.read_csv(
    "savant_2024_paul skenes(pitch type).csv"
)

savant_pitch_type_2025 = pd.read_csv(
    "savant_2025_paul skenes(pitch type).csv"
)
```

```python
# 2024年Referenceデータをテーブルに登録
reference_2024.to_sql(
    "reference_2024",
    conn,
    if_exists="replace",
    index=False
)

# 2025年Referenceデータをテーブルに登録
reference_2025.to_sql(
    "reference_2025",
    conn,
    if_exists="replace",
    index=False
)

# 2024年Savantデータをテーブルに登録
savant_2024.to_sql(
    "savant_2024",
    conn,
    if_exists="replace",
    index=False
)

# 2025年Savantデータをテーブルに登録
savant_2025.to_sql(
    "savant_2025",
    conn,
    if_exists="replace",
    index=False
)

# 2024年球種別Savantデータをテーブルに登録
savant_pitch_type_2024.to_sql(
    "savant_pitch_type_2024",
    conn,
    if_exists="replace",
    index=False
)

# 2025年球種別Savantデータをテーブルに登録
savant_pitch_type_2025.to_sql(
    "savant_pitch_type_2025",
    conn,
    if_exists="replace",
    index=False
)
```

```python
# 2025年の主要成績を取得
query = """
SELECT
    GS AS 先発登板数,
    W AS 勝利,
    L AS 敗戦,
    IP AS 投球回,
    ERA AS 防御率,
    SO AS 奪三振,
    WHIP,
    WAR
FROM reference_2025;
"""

# SQLを実行してデータを取得
main_stats_2025 = pd.read_sql_query(query, conn)

main_stats_2025
```

```python
# 2024年・2025年の主要成績を比較
query = """
SELECT
    '2024' AS 年,
    IP AS 投球回,
    ERA AS 防御率,
    SO AS 奪三振,
    WHIP,
    WAR
FROM reference_2024

UNION ALL

SELECT
    '2025' AS 年,
    IP AS 投球回,
    ERA AS 防御率,
    SO AS 奪三振,
    WHIP,
    WAR
FROM reference_2025;
"""

comparison_stats = pd.read_sql_query(query, conn)

comparison_stats
```

```python
# 2024年・2025年のK%・BB%を取得
query = """
SELECT
    '2024' AS 年,
    k_percent AS K率,
    bb_percent AS BB率
FROM savant_2024

UNION ALL

SELECT
    '2025' AS 年,
    k_percent AS K率,
    bb_percent AS BB率
FROM savant_2025;
"""

rate_stats = pd.read_sql_query(query, conn)

rate_stats
```

```python
# 主要成績とK%・BB%を結合
comparison_stats = comparison_stats.merge(
    rate_stats,
    on="年"
)

comparison_stats
```
